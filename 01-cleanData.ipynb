```python
from sklearn.preprocessing import MinMaxScaler, normalize, StandardScaler, Normalizer
import pandas as pd
```

```python
df = pd.read_csv('train-v1.2.csv')
```

```python
cols = df.columns[0:-1]
# X = df[cols]
# y = df['class information']
```

```python
x_cols = df.columns[1:-1]
x_cols
```

```python
std_scaler = StandardScaler()
df_scaled = df.copy()
df_scaled[x_cols] = std_scaler.fit_transform(df_scaled[x_cols])
```

```python
df_scaled = df_scaled.groupby('id').mean()
df_scaled['class information'] = df_scaled['class information'].apply(int)
df_scaled.head()
```

```python
normalizer = Normalizer()
df_norm = df.copy()
df_norm[x_cols] = normalizer.fit_transform(df_norm[x_cols])
```

```python
df_norm = df_norm.groupby('id').mean()
df_norm['class information'] = df_norm['class information'].apply(int)
df_norm.head()
```
