```python
from sklearn.preprocessing import MinMaxScaler, normalize, StandardScaler, Normalizer
import pandas as pd
```

```python
df = pd.read_csv('train-v1.2.csv')
```

```python
df.describe()
```

```python
# X.head()
```

## **Columnas**

```python
x_cols = df.columns[1:-1]
x_cols
```

## Standard Scaler

```python
std_scaler = StandardScaler()
df_scaled = df.copy()
df_scaled[x_cols] = std_scaler.fit_transform(df_scaled[x_cols])
```

```python
df_scaled.head()
```

```python
df_scaled.describe()
```

```python
# df_scaled = pd.DataFrame(df_scaled, columns=df.columns)
# df_scaled.head()
```

```python
# df_scaled.head()
```

```python
df_scaled = df_scaled.groupby('id').mean()
df_scaled['class information'] = df_scaled['class information'].apply(int)
df_scaled.head()
```

## Normalizer

```python
normalizer = Normalizer()
df_norm = df.copy()
df_norm[x_cols] = normalizer.fit_transform(df_norm[x_cols])
```

```python
df_norm = df_norm.groupby('id').mean()
df_norm['class information'] = df_norm['class information'].apply(int)
df_norm.head()
```

```python
df_norm[x_cols].describe()
```

## Normalizacion por MinMax

```python
minmax_scaler = MinMaxScaler()
df_minmax = df.copy()
```

```python
df_minmax[x_cols] = minmax_scaler.fit_transform(df_minmax[x_cols])
```

```python
df_minmax = df_minmax.groupby('id').mean()
df_minmax['class information'] = df_minmax['class information'].apply(int)
df_minmax.head()
```

```python
df_minmax[x_cols].describe()
```

```python
df_minmax.to_csv('train_minmax_scaled.csv')
df_norm.to_csv('train_norm_scaled.csv')
df_scaled.to_csv('train_scaled.csv')
```
